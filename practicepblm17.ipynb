```python
print("Pattern-1",end="\t")
print("Right sliding")
n=5
for i in range(n+1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(i):
        print("*",end="")
    print()
print()


print("Patter-2",end="\t")
print("left sliding")
for i in range(n+1):
    for j in range(n-i):
        print("*",end="")
    # for k in range(i):
    #     print("*",end="")
    print()
print()



print()
print("Pattern-3",end="\t")
print("Triangular pattern starting with 2")
for i in range(1,n+1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i):
        print("*",end="")
    print()

print()
print("Pattern-4",end="\t")
print("Triangular pattern starting with 1")
for i in range(1,n+1):
    
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()
print()

print("Pattern-5",end="\t")
print("Diamond")
for i in range(1,n+1):
    
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()
for i in range(n,0,-1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    
    print()
```
