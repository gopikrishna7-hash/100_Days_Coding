```python
if __name__ == '__main__':
    N = int(input())

    list_a = []
    i = 0

    while i < N:
        inpt = input().split()

        match inpt[0]:
            case "append":
                idx_1 = int(inpt[1])
                list_a.append(idx_1)

            case "insert":
                idx_1 = int(inpt[1])
                idx_2 = int(inpt[2])
                list_a.insert(idx_1, idx_2)

            case "remove":
                idx_1 = int(inpt[1])
                list_a.remove(idx_1)

            case "pop":
                list_a.pop(-1)

            case "reverse":
                list_a.reverse()

            case "sort":
                list_a.sort()

            case "print":
                print(list_a)

        i += 1
```
