```python
# dict Comprehension
dict_cmp={ x: x**2 for x in range(1,7)}
print(dict_cmp[2])
```

```python
l_tup=[("hi",23),("yes",12),("yeah",44)] # list of tup to dict
d1=dict(l_tup)
print(d1)

l_tup1=[["hi",23],["yes",12],["yeah",44]] # list of list to dict
d2=dict(l_tup1)
print(d2)

tup_tup=(("hi",23),("yes",12),("yeah",44)) # tup of tup to dict
d3=dict(tup_tup)
print(d3)

tup_l=(["hi",23],["yes",12],["yeah",44]) # tup of list to dict
d4=dict(tup_l)
print(d4)

# tup_l={[2,4]}                              # set of list error bec sets won't store Mutable items
# set_l=dict(tup_l)
# print(set_l)
```

# Accessing dict items

```python
d_items={
    "th":40,
    "sp":50,
    "plp":25,
    "app":35,
    "wtr":60
}
```

```python
val=d_items["app"]        
val1=d_items.get("th")

print(val)
print(val1)
print(d_items.get("lk","key not present"))

```

```python
for i in d_items:
    print(i,d_items[i])
print(d_items.keys())
print(d_items.values())
print(d_items.items())
```

```python
dr_items={
    "th":40,
    "cola_items":{
        "sp":50,
        "plp":25,
        "app":35,
        "wtr":60
    }
}
```

```python
print(dr_items["cola_items"]["plp"])
print(dr_items["cola_items"].get("app"))
```

```python
# Insert & Update Operations Indirectly

x_dict={"name": "Ch", "age":23}
x_dict["age"]=25                 # Update
x_dict["city"]="Hyd"             # Insert
print(x_dict)
```

```python
# Insert & Update Operations directly

x_dict.update({"age": 24})             # Update
x_dict.update({"age": 25,'Occ':"doc", "Salary":1.5})             # Insert # Update
print(x_dict)
```

# Deleting items from list pop(), popitem(),clear(),del()

```python
x_dict.pop("city")
print(x_dict)
```

```python
x_dict.popitems()
print(x_dict)
```

```python
x_dict.clear()
print(x_dict)
```

```python
del x_dict
print(x_dict)
```
